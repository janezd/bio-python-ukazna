```python
weight = int(input("Body weight: "))
height = float(input("Body height: "))
if height > 100:
    height /= 100
BMI = weight / height ** 2
print("BMI: ", BMI)
if BMI > 24:
    print("Perhaps a bit too much")
elif BMI < 18:
    print("Perhaps you should gain some weight (or decrease your height, if possible)")
else:
    print("Ideal.")
```
