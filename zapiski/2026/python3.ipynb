## Overlapping k-mers

```python
k = 3
kmer_freq = {}
for inx in range(0, len(seq)-k+1):
    subseq = seq[inx:inx+k]
    kmer_freq[subseq] = kmer_freq.get(subseq, 0) + 1
```

```python
kmer_freq
```

Pack into a function `kmer_count(seq, k)` and returns a dictionary of frequencies of `k-mers` in `seq`.

```python
def kmer_count(seq, k=3): # k=3 is the default
    kmer_freq = {}
    for inx in range(0, len(seq)-k+1):
        subseq = seq[inx:inx+k]
        kmer_freq[subseq] = kmer_freq.get(subseq, 0) + 1
    return kmer_freq
```

```python
kmer_count("ATATATATATCGCGATAT", 5)
```

```python
kmer_count("ATATATATATCGCGATAT")
```

```python
kmer_freq = kmer_count("ATATATATATCGCGATTATA")
```

```python
kmer_freq
```

```python
def get_max_key(dd):
    max_value = None
    max_key = None
    for k, v in dd.items():
        if max_value is None:
            max_value = v
            max_key = k
        elif v > max_value:
            max_value = v
            max_key = k
    return max_value, max_key
        
```

```python
def get_max_key_all(dd):
    max_value = None
    max_key = []
    for k, v in dd.items():
        if max_value is None:
            max_value = v
            max_key = [k]
        elif v > max_value:
            max_value = v
            max_key = [k]
        elif v == max_value:
            max_key.append(k)
    return max_value, max_key
        
```
